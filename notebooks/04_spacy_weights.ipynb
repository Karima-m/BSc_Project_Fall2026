## Methods

```python
"""
Method for creating NER dictionaries from wikipedia bio

Refrences: 
https://ricardolezama.com/linguistics/frequency-counts-for-named-entities/
https://stackoverflow.com/questions/71590119/how-to-count-the-occurances-of-all-entities-in-a-python-spacy-document


The two different pipelines for entitiy recognition which were considered

Dictionary creation method using en_core_web_lg
"""

def dictCreateLG(bio):
    
    nlp_accuracy = spacy.load('en_core_web_lg')

    ents = Counter()

    for ent in nlp_accuracy(bio).ents:
        ents[f"{ent.label_}:{ent.text}"] += 1
    return ents
```

```python
"""
Dictionary creation method using the trf spaCy pipeline
"""
def dictCreateTRF(bio):
    
    nlp_accuracy = spacy.load('en_core_web_trf')

    ents = Counter()

    for ent in nlp_accuracy(bio).ents:
        ents[f"{ent.label_}:{ent.text}"] += 1
    return ents
```

```python
"""
Dictionary creation method using the sm spaCy pipeline
"""
def dictCreateSM(bio):
    
    nlp_accuracy = spacy.load('en_core_web_sm')

    ents = Counter()

    for ent in nlp_accuracy(bio).ents:
        ents[f"{ent.label_}:{ent.text}"] += 1
    return ents
```

```python
# Reggex pattern to match for Napoleon
pattern = re.compile(r'^[A-Z_]+:Napol[eé]on( Bonaparte| I)?$')
```

```python
# One count per entity regardless of tag

def combine_ents(ner_counter):
    texts = Counter()
    for key, n in ner_counter.items():
        label, text = key.split(':', 1)   # split at the first ':'
        texts[text.strip()] += n
    return texts

LG_ents = combine_ents(LG_napoleon_weights)
TRF_ents = combine_ents(TRF_napoleon_weights)
```

```python
ROMAN = re.compile(r'^[IVXLC]+$')

def ents_by_text(ner_counter):
    """Sum counts per entity text across all labels."""
    texts = Counter()
    for key, n in ner_counter.items():
        label, text = key.split(':', 1)
        texts[text.strip()] += n
    return texts

def short_name(title):
    """Clean title: remove (...), text after comma, and ' of ...'. Keeps numerals."""
    name = re.sub(r'\s*\(.*?\)', '', title)
    name = name.split(',')[0]
    name = name.split(' of ')[0]
    return name.strip()

def name_parts(short):
    """Words used for first/last split, with Roman numerals removed."""
    return [w for w in short.split() if not ROMAN.match(w)]

def count_mentions(ent_texts, names_df):
    rows = []
    for _, row in names_df.iterrows():
        full = row['page_title']
        short = short_name(full)
        words = name_parts(short)

        # Full match: raw title + cleaned title + cleaned title without numerals
        no_numeral = ' '.join(words)
        full_variants = {full}
        if short != full and len(short.split()) > 1:
            full_variants.add(short)              # "Wenceslaus IV"
        if no_numeral != short and no_numeral:
            full_variants.add(no_numeral)         # "Wenceslaus"
        full_count = sum(ent_texts[v] for v in full_variants)

        # First and last name: only if more than one word after cleaning
        first_count = last_count = 0
        if len(words) > 1:
            first_count = ent_texts[words[0]]
            last_count = ent_texts[words[-1]]

        rows.append({
            'Name': full,
            'ID': row['wikidata_id'],
            'variants': sorted(full_variants),
            'parts': [words[0], words[-1]] if len(words) > 1 else [],
            'full_count': full_count,
            'first_count': first_count,
            'last_count': last_count,
            'weight': full_count + first_count + last_count,
        })
    return pd.DataFrame(rows).sort_values('weight', ascending=False)

LG_napoleon_targets = count_mentions(ents_by_text(LG_napoleon_weights), names)
TRF_napoleon_targets = count_mentions(ents_by_text(TRF_napoleon_weights), names)
```
